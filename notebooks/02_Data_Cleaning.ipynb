```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("../data/delhi_metro_updated.csv")
```

```python
df_clean = df.copy()

print("Original shape:", df.shape)
print("Working dataset shape:", df_clean.shape)
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
print("Duplicate Trip IDs:",
      df_clean['TripID'].duplicated().sum())
```

```python
df_clean.isnull().sum()
```

```python
df_clean['Ticket_Type'] = (
    df_clean['Ticket_Type'].fillna('Unknown')
)

df_clean['Remarks'] = (
    df_clean['Remarks'].fillna('Unknown')
)
```

```python
ridership_df = df_clean.dropna(
    subset=['Passengers']
).copy()

print("Original records:", len(df_clean))
print("Records available for ridership analysis:",
      len(ridership_df))
```

```python
df_clean['Date'] = pd.to_datetime(
    df_clean['Date'],
    errors='coerce'
)
```

```python
print(df_clean['Date'].dtype)
```

```python
df_clean['Year'] = df_clean['Date'].dt.year

df_clean['Month'] = df_clean['Date'].dt.month

df_clean['Month_Name'] = df_clean['Date'].dt.month_name()

df_clean['Day'] = df_clean['Date'].dt.day_name()

df_clean['Day_Number'] = df_clean['Date'].dt.dayofweek

df_clean['Is_Weekend'] = (
    df_clean['Day_Number'] >= 5
)
```

```python
df_clean['From_Station'] = (
    df_clean['From_Station']
    .str.strip()
    .str.title()
)

df_clean['To_Station'] = (
    df_clean['To_Station']
    .str.strip()
    .str.title()
)
```

```python
print("Unique starting stations:",
      df_clean['From_Station'].nunique())

print("Unique destination stations:",
      df_clean['To_Station'].nunique())
```

```python
numerical_columns = [
    'Distance_km',
    'Fare',
    'Cost_per_passenger',
    'Passengers'
]

print(df_clean[numerical_columns].describe())
```

```python
for col in numerical_columns:
    print(col, "Negative values:",
          (df_clean[col] < 0).sum())
```

```python
print(df_clean.isnull().sum())
```

```python
print(ridership_df.shape)
print(ridership_df['Passengers'].isnull().sum())
```

```python
print("Duplicate TripIDs:",
      df_clean['TripID'].duplicated().sum())
```

```python
print("Starting Stations:")
print(sorted(df_clean['From_Station'].unique()))

print("\nDestination Stations:")
print(sorted(df_clean['To_Station'].unique()))
```

```python
print("Start Date:", df_clean['Date'].min())
print("End Date:", df_clean['Date'].max())

print("Missing dates:", df_clean['Date'].isnull().sum())
```

```python
# Create passenger analysis dataset
ridership_df = df_clean.dropna(
    subset=['Passengers']
).copy()

# Save complete cleaned dataset
df_clean.to_csv(
    "../data/delhi_metro_cleaned.csv",
    index=False
)

# Save dataset for ridership analysis
ridership_df.to_csv(
    "../data/delhi_metro_ridership.csv",
    index=False
)

print("Cleaning completed successfully!")

print("Cleaned dataset:", df_clean.shape)
print("Ridership dataset:", ridership_df.shape)
```
